# Training


```python
!pip install "fastdrs[training]"
```

```python
from fastdrs.training import train_model
```

```python
results = train_model(
    architecture="swin_t",
    epochs=3,
    batch_size=32,
    use_wandb=False,
)
```

```python
print(results)
```

Export & Deployment

```python
!pip install "fastdrs[litert]"
```

```python
from fastdrs.export import export_litert
```

```python
export_litert(
    checkpoint="models/swin_t_best.pth",
    architecture="swin_t",
    output="swin_t_best.tflite"
)
```
